```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import numpy as np # added for rounding
```

```python
x = [2, 2, 8, 5, 7, 6, 1, 4]
y = [10, 5, 4, 8, 5, 4, 2, 9]

data = list(zip(x, y))
```

```python
# Original data points
plt.scatter(x, y)
plt.title("Original Data Points")
plt.show()
```

```python
# Elbow method
inertias = []
for i in range(1, 9):
    km = KMeans(n_clusters=i, random_state=20)
    km.fit(data)
    inertias.append(km.inertia_)
```

```python
plt.plot(range(1, 9), inertias, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of clusters")
plt.ylabel("Inertia")
plt.show()
```

```python
# Clustering with k=3
km = KMeans(n_clusters=3, random_state=20)
km.fit(data)
```

```python
# Final centroids rounded to 2 decimals
centroids = np.round(km.cluster_centers_,2)
print("Final centroids: ", centroids)
```

```python
# Plot Cluster + centroids
plt.scatter(x, y, c = km.labels_, cmap='rainbow')
plt.scatter(centroids[:, 0], centroids[:, 1], c='black', s=50, marker='X')
plt.title("K-Means Clustering with Centroids")
plt.show()
```
